### 1) Recent Trends in IoT (Lecture)  

>**Key Action**: Understand the concept of IoT and recent trends in IoT.

**1. What is IoT(Internet of Things)?**
- A network of physical devices, vehicles, appliances, and other physical objects that are embedded with sensors, software, and network connectivity, allowing them to collect and share data (IBM)
- Thing/Device → IoT Gateway (filtering data) → Cloud → User Interface (e.g., IoT app, dashboard)
- Example Application Areas:
  + managing traffic patterns with smart cars and othre smart automotive devices
  + monitoring environmental conditions in farms
  + tracking inventory and shipments in warehouses
  + optimizing comfort, security, and energy use in smart homes (e.g., smart thermostats, lighting, and plugs)

**2. AIoT (AI+IoT)**
- This integration creates smart, connected systems that collect data from IoT devices and use AI to analyze that data, enabling intelligent decision-making without direct human intervention
- IoT delivers value through improved connectivity, signaling, and data exchange, while AI builds on this foundation by enhancing data analytics and predictions, optimizing operation, and enabling more advanced decision-making capabilities

**Example: John Deere's Smart Farming**
- A global agricultural company leading innovation in agricultural, turf, construction, and forestry equipment
- Autonomous tractors: use AIoT for 360-degree obstacle detection and real-time decision-making, allowing driveless operation and helping address labor shortages
- See & Spray: An AIoT-enabled sprayer that uses boom-mounted cameras and onboard AI to detect weeds in real time and spray only where needed, reducing herbicide use by up to 80 percent

**Example: Amazon's Smart Logistics**
- Robotics Fulfillment Center: A scalable AIoT ecosystem that orchestrates over one million robots using the "DeepFleet" GenAI foundation model. This enables real-time autonomous routing and optimize warehouse traffic
- Amazon Prime Air: an AIoT drone delivery service that connects drones and ground sensors to the cloud. AI plans safe routes and avoids obstacles in real time, reducing human intervention and accelerating delivery

**Example: Samsung's SamrtThings**
- A Samsung online IoT platform that allows users to control multiple devices under one single app
- Home Insight: uses AI and data from connected home devices to continuously monitor the home environment. The system learns daily usage patterns and provides at-a-glance view of personalized recommendations
- AI Energy Mode: uses AI and connected appliancees to monitor energy usage in real time and automatically switch to energy-saving settings. This can reduce electricity consumption by approximately 30 percent annually
- Routine Creation Assistant (with Generative AI): allows users to describe the automation routine they want in natural language. Based on device-usage patterns, the systems recommends suitable automations, removing the need to manually select each device, condition and action
